## Requisitos y wireframes UX
Notebook complementario. **No reemplaza el piloto original.** Datos sintéticos o ilustrativos; no constituyen validación real. Ejecutar en Colab/Jupyter y descargar la carpeta de salida.


```python
import pandas as pd
from pathlib import Path
out=Path("salida_ind_sis02");out.mkdir(exist_ok=True)
perfiles=pd.DataFrame([{"perfil":"Tecnico","tarea":"analizar causas y priorizar riesgos"},{"perfil":"Operativo","tarea":"identificar acciones inmediatas"}])
req=pd.DataFrame([{"id":f"RF-{i:02d}","requisito":t} for i,t in enumerate(["mostrar mapa de zonas","filtrar por periodo","consultar H-E-V de zona","ordenar ranking de criticidad","exportar resumen"],1)])
ux=pd.DataFrame([{"id":f"UX-{i:02d}","criterio":t} for i,t in enumerate(["etiquetas comprensibles","contraste legible","consistencia de navegacion","retroalimentacion de filtros","accesibilidad por teclado"],1)])
for n,d in [("perfiles",perfiles),("requisitos",req),("usabilidad",ux)]:d.to_csv(out/f"{n}.csv",index=False)
(out/"wireframe_mapa.txt").write_text("[RISKTWIN INDUSTRIAL]\n[Filtros: fecha | tipo de riesgo]\n[MAPA: ZONAS A/B/C] [Ranking de zonas]\n[Seleccionar zona -> detalle]\n",encoding="utf-8")
(out/"wireframe_detalle.txt").write_text("[Zona seleccionada: IND-01]\n[H: peligrosidad] [E: exposicion] [V: vulnerabilidad]\n[Eventos recientes]\n[Acciones preventivas propuestas]\n[Volver al mapa]\n",encoding="utf-8")
print("2 perfiles, 5 requisitos, 5 criterios y 2 wireframes de baja fidelidad.")
```
