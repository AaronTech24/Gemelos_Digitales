## Grafo pequeño de intersecciones
Notebook complementario. **No reemplaza el piloto original.** Datos sintéticos o ilustrativos; no constituyen validación real. Ejecutar en Colab/Jupyter y descargar la carpeta de salida.


```python
import networkx as nx, pandas as pd, matplotlib.pyplot as plt
from pathlib import Path
out=Path("salida_urb02");out.mkdir(exist_ok=True)
# Topología sintética; sustituir por geometría verificada de intersección en Vía Perimetral.
G=nx.Graph()
G.add_nodes_from([(f"N{i}",{"tipo":"interseccion","zona":"PER-01"}) for i in range(1,7)])
G.add_weighted_edges_from([("N1","N2",120),("N2","N3",90),("N3","N4",130),("N2","N5",75),("N5","N6",80),("N6","N4",110)],weight="longitud_m")
grado=dict(G.degree());inter=nx.betweenness_centrality(G,weight="longitud_m",normalized=True)
nodos=pd.DataFrame([{"id_nodo":n,"grado":grado[n],"intermediacion":inter[n]} for n in G])
aristas=pd.DataFrame([{"origen":u,"destino":v,"longitud_m":d["longitud_m"]} for u,v,d in G.edges(data=True)])
nodos.to_csv(out/"nodos.csv",index=False);aristas.to_csv(out/"aristas.csv",index=False)
nx.draw(G,with_labels=True,node_size=900);plt.title("Grafo ilustrativo, NO georreferenciado")
plt.savefig(out/"grafo_piloto.png",dpi=160,bbox_inches="tight");plt.show()
print("Métricas: grado (conexiones directas) e intermediación (participación en rutas mínimas).")
print(nodos.to_string(index=False))
```
