```python
import pandas as pd, json
from pathlib import Path
out=Path("salida_urb01");out.mkdir(exist_ok=True)
zonas=pd.DataFrame({"id_zona":[f"PER-{i:02d}" for i in range(1,6)],"nombre":[f"Zona piloto sintética {i}" for i in range(1,6)],"tipo_riesgo":["vial","inundacion","seguridad","vial","inundacion"],"periodo":["2026-01"]*5})
variables=[
("vial","H","siniestros","eventos/mes","ATM / registros institucionales"),
("vial","H","atropellamientos","eventos/mes","ATM / registros institucionales"),
("vial","H","velocidad_operativa","km/h","aforos / levantamiento"),
("vial","E","flujo_vehicular","vehiculos/h","ATM / aforos"),
("vial","E","flujo_peatonal","personas/h","levantamiento"),
("vial","V","iluminacion_deficiente","proporcion","inventario vial"),
("vial","V","ausencia_cruce_seguro","binario","inspeccion"),
("inundacion","H","precipitacion","mm/dia","INAMHI"),
("inundacion","H","anegamientos","eventos/año","municipio / ECU911"),
("inundacion","E","poblacion_expuesta","personas","INEC"),
("inundacion","E","infraestructura_expuesta","activos","municipio"),
("inundacion","V","drenaje_obstruido","proporcion","inspeccion municipal"),
("inundacion","V","cota_baja","m","modelo digital elevacion"),
("seguridad","H","robos","eventos/mes","Policia / ECU911"),
("seguridad","H","homicidios","eventos/mes","Policia / fiscalia"),
("seguridad","E","flujo_personas","personas/h","aforos"),
("seguridad","E","comercios_expuestos","establecimientos","catastro"),
("seguridad","V","iluminacion_insuficiente","proporcion","inspeccion"),
("seguridad","V","acceso_respuesta","minutos","instituciones competentes")]
dic=pd.DataFrame(variables,columns=["riesgo","dimension_HEV","variable","unidad","fuente_potencial"])
dic["tipo_dato"]="numerico";dic["estado"]="propuesta_por_verificar"
dic.to_csv(out/"diccionario_19_variables.csv",index=False)
zonas.to_csv(out/"zonas_piloto_sinteticas.csv",index=False)
# Geometrías ilustrativas: NO representan límites geográficos reales de la Perimetral.
features=[{"type":"Feature","properties":r.to_dict(),"geometry":None} for _,r in zonas.iterrows()]
(out/"zonas_piloto_sinteticas.geojson").write_text(json.dumps({"type":"FeatureCollection","features":features},ensure_ascii=False,indent=2),encoding="utf-8")
print("19 variables; 5 zonas sintéticas. Geometrías pendientes de delimitación real.")
```
