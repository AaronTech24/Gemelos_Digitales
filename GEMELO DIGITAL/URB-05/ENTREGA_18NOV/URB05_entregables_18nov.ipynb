## Tres escenarios preventivos
Notebook complementario. **No reemplaza el piloto original.** Datos sintéticos o ilustrativos; no constituyen validación real. Ejecutar en Colab/Jupyter y descargar la carpeta de salida.


```python
import pandas as pd
from pathlib import Path
out=Path("salida_urb05");out.mkdir(exist_ok=True)
# Prototipo determinista ilustrativo, NO predice siniestros reales.
tramos=pd.DataFrame({"id_segmento":["PER-S01","PER-S02","PER-S03","PER-S04","PER-S05"],"velocidad_kmh":[70,65,75,60,70],"flujo_veh_h":[650,720,500,800,680],"exposicion_peatonal":[.7,.5,.8,.4,.6],"separacion_flujos":[0,0,0,0,0]})
def proxy_riesgo(df):
 # Proxy pedagógico, no probabilidad calibrada: velocidad relativa x flujo relativo x exposición peatonal.
 return (df.velocidad_kmh/80)*(df.flujo_veh_h/1000)*df.exposicion_peatonal*(1-.5*df.separacion_flujos)
filas=[]
for nombre in ["base","reduccion_velocidad","separacion_flujos"]:
 d=tramos.copy()
 if nombre=="reduccion_velocidad":d["velocidad_kmh"]*=.8
 if nombre=="separacion_flujos":d["separacion_flujos"]=1
 d["proxy_riesgo"]=proxy_riesgo(d)
 filas.extend([{"escenario":nombre,**r.to_dict()} for _,r in d.iterrows()])
res=pd.DataFrame(filas);res.to_csv(out/"comparacion_escenarios.csv",index=False)
print(res.groupby("escenario")["proxy_riesgo"].mean().round(4))
print("Supuestos: velocidad -20%; separación reduce proxy 50%; efectos hipotéticos, no estimaciones empíricas.")
```
